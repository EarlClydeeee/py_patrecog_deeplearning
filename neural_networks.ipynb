## Part 1 – Importing Required Libraries

### Step 1: Import NumPy and Matplotlib

NumPy is used for mathematical operations and matrix computation. Matplotlib is used for graph plotting and visualization.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Part 2 – Understanding a Single Artificial Neuron

### Step 2: Create Inputs, Weights, and Bias

The neuron multiplies inputs by weights and adds a bias to generate an output.

```python
inputs = np.array([1, 2, 3])
weights = np.array([0.2, 0.8, -0.5])
bias = 2
weighted_sum = np.dot(inputs, weights) + bias
print("Weighted Sum:", weighted_sum)
```

## Part 3 – Activation Functions

### Step 3: Implement Sigmoid Function

The Sigmoid function converts values between 0 and 1.

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

### Step 4: Plot the Sigmoid Function

The graph visualizes how the Sigmoid activation function behaves.

```python
x = np.linspace(-10, 10, 100)
y = sigmoid(x)
plt.plot(x, y)
plt.title("Sigmoid Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.show()
```

### Step 5: Implement ReLU Function

ReLU removes negative values and keeps positive values unchanged.

```python
def relu(x):
    return np.maximum(0, x)
```

### Step 6: Plot the ReLU Function

The ReLU graph shows zero values for negative inputs and positive linear output for positive inputs.

```python
y = relu(x)
plt.plot(x, y)
plt.title("ReLU Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.show()
```

## Part 4 – Building a Simple Neural Network

### Step 7: Create a Hidden Layer

Forward propagation computes outputs from one layer to another.

```python
X = np.array([[1, 2, 3]])
weights1 = np.array([
    [0.2, 0.8],
    [-0.5, 0.1],
    [0.3, -0.9]
])
bias1 = np.array([0.1, 0.2])
hidden_layer = np.dot(X, weights1) + bias1
print(hidden_layer)
```

### Step 8: Apply ReLU Activation

Activation functions allow neural networks to learn complex patterns.

```python
hidden_output = relu(hidden_layer)
print(hidden_output)
```

## Part 5 – Loss Function

### Step 9: Compute Mean Squared Error

Loss functions measure how far predictions are from actual outputs.

```python
y_true = np.array([1, 0, 1])
y_pred = np.array([0.8, 0.2, 0.9])
mse = np.mean((y_true - y_pred) ** 2)
print("MSE:", mse)
```

## Part 7 – MNIST Handwritten Digit Classification

### Step 12: Load the Dataset

The MNIST dataset contains handwritten digits from 0 to 9.

```python
mnist = keras.datasets.mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()
print(X_train.shape)
```

### Step 13: Display Sample Image

The dataset image is displayed to verify the handwritten digit.

```python
plt.imshow(X_train[0], cmap='gray')
plt.title(f"Label: {y_train[0]}")
plt.show()
```

### Step 14: Normalize the Dataset

Normalization scales pixel values between 0 and 1.

```python
X_train = X_train / 255.0
X_test = X_test / 255.0
```

## Part 8 – Building a Deep Neural Network

### Step 15: Create the Neural Network Model

The Flatten layer converts the image into a one-dimensional vector. Dense layers perform classification.

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])
model.summary()
```

### Step 16: Compile the Model

Compilation configures the optimizer, loss function, and evaluation metric.

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

### Step 17: Train the Model

Training allows the model to learn from data through multiple epochs.

```python
history = model.fit(
    X_train,
    y_train,
    epochs=5
)
```

### Step 18: Evaluate the Model

Evaluation measures the performance of the trained model using unseen data.

```python
test_loss, test_acc = model.evaluate(X_test, y_test)
print("Test Accuracy:", test_acc)
```

### Step 19: Make Predictions

The model predicts the handwritten digit based on learned patterns.

```python
predictions = model.predict(X_test)
print("Predicted Class:", np.argmax(predictions[0]))
print("Actual Class:", y_test[0])
```

## Part 9 – Convolutional Neural Network (CNN)

### Step 20: Create CNN Model

CNN models are specialized neural networks used for image processing.

```python
cnn_model = keras.Sequential([
    keras.layers.Reshape((28, 28, 1), input_shape=(28, 28)),
    keras.layers.Conv2D(32, (3, 3), activation='relu'),
    keras.layers.MaxPooling2D((2, 2)),
    keras.layers.Flatten(),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])
cnn_model.summary()
```

### Step 21: Compile and Train CNN Model

CNN models usually perform better in image classification tasks.

```python
cnn_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
cnn_model.fit(X_train, y_train, epochs=3)
```
